```python
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance
```

```python
input_path = "/content/PlantVillage"
output_path = "/content/ProcessedData"
```

```python
# import os
# os.listdir('/content/drive/MyDrive')
```

```python
if not os.path.exists(output_path):
    os.makedirs(output_path)
```

```python
#step5
def process_image(img_path):
    img = cv2.imread(img_path)

    if img is None:
        return None

    # Convert BGR → RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Resize
    img = cv2.resize(img, (224, 224))

    # Convert to PIL
    img_pil = Image.fromarray(img)

    # Contrast
    contrast = ImageEnhance.Contrast(img_pil)
    img_pil = contrast.enhance(1.2)

    # Brightness
    brightness = ImageEnhance.Brightness(img_pil)
    img_pil = brightness.enhance(1.1)

    # Convert back to numpy
    img = np.array(img_pil)

   #image sharpening
    kernel = [[0,-1,0],[-1,5,-1],[0,-1,0]]
    kernel = np.array(kernel)
    img = cv2.filter2D(img, -1, kernel)

    return img
```

```python
#step 6
count = 0

for split in ["train", "val"]:

    print("Processing:", split)

    split_input = os.path.join(input_path, split)
    split_output = os.path.join(output_path, split)

    if not os.path.exists(split_output):
        os.makedirs(split_output)

    for class_name in os.listdir(split_input):

        class_input = os.path.join(split_input, class_name)
        class_output = os.path.join(split_output, class_name)

        if not os.path.exists(class_output):
            os.makedirs(class_output)

        for img_name in os.listdir(class_input):

            img_path = os.path.join(class_input, img_name)

            processed_img = process_image(img_path)

            if processed_img is not None:
                save_path = os.path.join(class_output, img_name)

                #convert back to rgb
                processed_img = cv2.cvtColor(processed_img, cv2.COLOR_RGB2BGR)

                cv2.imwrite(save_path, processed_img)

                count += 1
                if count % 500 == 0:
                    print("Processed images:", count)

            else:
                print("Skipped:", img_path)
```

```python
for split in ["train", "val"]:
    print("\nChecking:", split)

    input_split = os.path.join(input_path, split)
    output_split = os.path.join(output_path, split)

    for folder in os.listdir(input_split):
        in_count = len(os.listdir(os.path.join(input_split, folder)))
        out_count = len(os.listdir(os.path.join(output_split, folder)))

        print(folder, "Input:", in_count, "Output:", out_count)
```
